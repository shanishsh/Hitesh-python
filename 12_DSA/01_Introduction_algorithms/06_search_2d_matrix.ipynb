```python
arr= [
    [1,3,5,7],
    [10,11,16,20],
    [23,30,34,60]
    ]

i=3
j=4
target=-10
row_idx=-1
col_idx=-1

for i in range(i-1,-1,-1):
    if arr[i][0]<=target:
        row_idx=i
        break
    
if row_idx==-1:
    for j in range(j):
        if arr[row_idx][j]==target:
            col_idx=j
            break


print(row_idx,col_idx)
```

```python
row=3
col=4
left=0
right=row*col-1
target=24
present=True

while(left<=right):
    mid=left+(right-left)//2
    row=mid//4
    col=mid%4

    if target==arr[row][col]:
        present=True
        break

    elif target>arr[row][col]:
        left=mid+1
    else:
        right=mid-1

    present=False

print(present)



```
