```python
import time
import multiprocessing

# --- 1. Standard Sequential Search (O(n) - Single Thread) ---
def sequential_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# --- 2. Parallel Search (O(n) - 4 Threads) ---
def scan_chunk(arr_chunk, target, start_idx):
    for i, val in enumerate(arr_chunk):
        if val == target:
            return start_idx + i
    return -1

def parallel_search(arr, target):
    n = len(arr)
    chunk_size = n // 4
    
    chunks = [
        (arr[0:chunk_size], target, 0),
        (arr[chunk_size:chunk_size*2], target, chunk_size),
        (arr[chunk_size*2:chunk_size*3], target, chunk_size*2),
        (arr[chunk_size*3:], target, chunk_size*3)
    ]
    
    with multiprocessing.Pool(processes=4) as pool:
        results = pool.starmap(scan_chunk, chunks)
        
    for index in results:
        if index != -1:
            return index
    return -1

# --- Benchmarking Setup ---
if __name__ == "__main__":
    print("Generating 20 million numbers...")
    # Create a large list: 0, 1, 2, ... up to 20,000,000
    large_array = list(range(20000000))
    
    # We choose a target near the very end so the algorithms must scan almost everything
    target_value = 19999990
    
    print(f"Searching for target: {target_value}\n")

    # Benchmark 1: Sequential
    start_time = time.perf_counter()
    seq_result = sequential_search(large_array, target_value)
    end_time = time.perf_counter()
    seq_duration = end_time - start_time
    print(f"[Sequential] Found at index: {seq_result}")
    print(f"[Sequential] Time taken: {seq_duration:.4f} seconds\n")

    # Benchmark 2: Parallel (4 CPU Cores)
    start_time = time.perf_counter()
    par_result = parallel_search(large_array, target_value)
    end_time = time.perf_counter()
    par_duration = end_time - start_time
    print(f"[Parallel]   Found at index: {par_result}")
    print(f"[Parallel]   Time taken: {par_duration:.4f} seconds\n")
    
    # Calculate the real-world improvement
    speedup = seq_duration / par_duration
    print(f"🚀 The 4-pointer parallel search was {speedup:.2f}x faster!")
```
